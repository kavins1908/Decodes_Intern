```python
import pandas as pd

df = pd.read_excel("Raw_Data_Cleaning_Project.xlsx")

df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df[df.isnull().any(axis=1)]
```

```python
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
```

```python
df["Age"].fillna(df["Age"].median(), inplace=True)
```

```python
df["City"].fillna(df["City"].mode()[0], inplace=True)
```

```python
df["Join_Date"].fillna(method="ffill", inplace=True)
```

```python
df.isnull().sum()
```

```python
df["Salary"].fillna(df["Salary"].median(), inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df[df.duplicated(subset="Employee_ID", keep=False)]
```

```python
df.drop_duplicates(subset="Employee_ID", keep="first", inplace=True)
```

```python
df["Employee_ID"].duplicated().sum()
```

```python
df["Name"] = df["Name"].str.strip()
df["City"] = df["City"].str.strip()
```

```python
df["Name"] = df["Name"].str.title()
```

```python
df["City"] = df["City"].str.title()
```

```python
df["Salary"] = df["Salary"].round(2)
```

```python
df.to_excel("Cleaned_Data.xlsx", index=False)
```

```python
df.head(10)
```

```python
df["Join_Date"] = pd.to_datetime(df["Join_Date"], format="mixed", dayfirst=True, errors="coerce")
```

```python
df["Join_Date"] = df["Join_Date"].dt.strftime("%Y-%m-%d")
```
